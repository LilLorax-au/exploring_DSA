### Queues
A Queue is one of the fundamental data structures used within computer science. The queue operates of the rule that what ever is placed in first must also come out first(FIFO). You can think of a queue simular to how you would line up to buy tickets for a show; whoever joined the line first, will be the first to be served, followed by who was after them, and continuing in this order, and only in this order.

Some more common use cases for Queue could include, but not limited to:
* Round Robin algorithms, in the planning and scheduling of tasks such as CPU processes and network load balancing
* Structured syntax handling, e.g. were brackets and parenthesis should open and close

#### Implementation
There is a few different ways Queues may be implemented, such as; Standard Queues, Circular Queues and the more distant Deque(_Pronounced deck_ also known as a double ended Queue). The example that is demonstrated below is a Circular Queue, implemented using a Dynamic Array.

First is the Queue as Abstract Data Type(AST), this is how we can define base behavior of queue, but can be modified for use case. Python doesn't support native interfaces, so a class with pass method is used to demonstration this. When going to use queue and queue-like structures, we will use this as a super/ parent class.

```python
from typing import Any

class QueueInterface:
    def enqueue(self, element: Any) -> None:
        """Insert/Add an element to the back of the queue"""
        pass
    def dequeue(self) -> Any:
        """Remove and return the element from the front of the queue"""
        pass
    def first(self) -> Any:
        """Safe return the first element in the queue"""
        pass
    def __len__(self) -> int:
        """Return the number of elements in the queue"""
        pass
    def is_empty(self) -> bool:
        """Return whether the queue is empty"""
        pass
```

Now we will implement the CircularArrayQueue using the above class

```python
from typing import Any

class CircularArrayQueue(QueueInterface):
    DEF_CAP: int = 10

    def __init__(self):
        """Initialize an empty queue"""
        # Define the inter structure, using a list as a Dynamic Array
        self.data: list = [None] * self.DEF_CAP
        self.size: int = 0
        self.front: int = 0


    def enqueue(self, element: Any) -> None:
        if self.size == len(self.data):
            self.resize(len(self.data) * 2)
        avail = (self.front + self.size) % len(self.data)
        self.data[avail] = element
        self.size += 1


    def dequeue(self) -> Any:
        if self.is_empty():
            raise IndexError("Queue is empty")

        value = self.data[self.front]
        self.data[self.front] = None
        self.front = (self.front + 1) % len(self.data)
        self.size -= 1
        return value

    def resize(self, cap: int) -> None:
        old = self.data
        self.data = [None] * cap
        walk = self.front

        for k in range(self.size):
            self.data[k] = old[walk]
            walk = (1 + walk) % len(old)
        self.front = 0

    def first(self) -> Any:
        if self.is_empty():
            raise IndexError("Queue is empty")
        else:
            return self.data[self.front]

    def __len__(self) -> int:
        return self.size

    def is_empty(self) -> bool:
        if self.size == 0:
            return True
        else:
            return False
```
