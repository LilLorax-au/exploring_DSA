### Queues
A Queue is one of the fundamental data structures used within computer science. The queue operates of the rule that whatever is placed in first must also come out first(FIFO). You can think of a queue simular to how you would line up to buy tickets for a show; whoever joined the line first, will be the first to be served, followed by who was after them, and continuing in this order, and only in this order.

Some more common use cases for Queue could include, but not limited to:
* Round Robin algorithms, in the planning and scheduling of tasks such as CPU processes and network load balancing
* Structured syntax handling, e.g. were brackets and parenthesis should open and close
* Used in Breadth-First search to traverse trees or graphs



#### Implementation
There is a few different ways Queues may be implemented, such as; Standard Queues, Circular Queues and the more distant Deque(_Pronounced deck_ also known as a double ended Queue). The example that is demonstrated below is a Circular Queue, implemented using a Dynamic Array.

First is the Queue as Abstract Data Type(AST), this is how we can define base behavior of queue, but can be modified for use case. Python doesn't support native interfaces, so a class with pass method is used to demonstration this. When going to use queue and queue-like structures, we will use this as a super/ parent class.

```python
from tabnanny import check
from typing import Any
from unittest import result


class EmptyQueue(Exception):
    """Exception raised when queue is empty"""
    pass


class QueueInterface:
    def enqueue(self, element: Any) -> None:
        """Insert/Add an element to the back of the queue"""
        pass
    def dequeue(self) -> Any:
        """Remove and return the element from the front of the queue"""
        pass
    def first(self) -> Any:
        """Safe return the first element in the queue"""
        pass
    def __len__(self) -> int:
        """Return the number of elements in the queue"""
        pass
    def is_empty(self) -> bool:
        """Return whether the queue is empty"""
        pass
```

Now we will implement the CircularArrayQueue using the above class

```python
from typing import Any

class CircularArrayQueue(QueueInterface):
    """Circular Array Queue, implemented with Queue ADT"""
    DEF_CAP: int = 10

    def __init__(self):
        """Initialize an empty queue"""
        # Define the inner structure, using a list as a Dynamic Array
        self.data: list = [None] * self.DEF_CAP
        self.size: int = 0
        self.front: int = 0


    def enqueue(self, element: Any) -> None:
        """Insert/Add an element to the back of the queue, has a time complexity of O(1) and maintains an atomized time complexity of O(1) on resize, only if resize doubles capacity"""
        # First check if the Array is full
        if self.size == len(self.data):
            # If at capacity, call resize
            self.resize(len(self.data) * 2)
        # Avail points to the back of the queue, this is found from: queues front index + size of the queue modulus the total size of array
        # E.G. if front sits at index 9, queue size of 4, and array size of 10, then avail points to index 3, (9 + 4) % 10 = 3
        # We do this to imutate the Array as if it were an ouroboros, its tail connecting to it head
        avail = (self.front + self.size) % len(self.data)
        # Now just add the element to where the new tail position sits
        self.data[avail] = element
        # Increment size
        self.size += 1


    def dequeue(self) -> Any:
        """Remove and return the element from the front of the queue, use case maintains a O(1) time complexity"""
        if self.is_empty():
            raise EmptyQueue("Queue is empty")

        # Set value to maintain a reference for the current front of the queue
        value = self.data[self.front]
        # Remove fronts value
        self.data[self.front] = None
        # Using simular process to find the back of the queue, we find the new front
        # Maintaining the circular structure, we use the last index of front, increment 1, and modulus the length of the Array,
        # E.G. if front index is 9, and Array length is 10, we get the new index of 0, (9 + 1) % 10 = 0
        self.front = (self.front + 1) % len(self.data)
        # decrement by 1
        self.size -= 1
        # return maintained old front
        return value

    def resize(self, cap: int) -> None:
        """Resize the queue to the specified capacity, maintains worse case time complexity of O(n)"""
        # Take a copy of the old data
        old = self.data
        # Create a new fixed sized list to mimic an Array, based on passed parameter
        self.data = [None] * cap
        # Grab a reference to the queues front
        walk = self.front

        # Iterate over all the old elements to place into new Array
        for k in range(self.size):
            # Place old front to the beginning of new Array
            self.data[k] = old[walk]
            # Find the next element in the old array using modulus arithmetic
            walk = (1 + walk) % len(old)
        # Reset front value to match new Array
        self.front = 0
        # Old Array leaves scope and is marked for garbage collection

    def first(self) -> Any:
        """Safe return the first element in the queue, Time complexity of O(1)"""
        # First check is Queue is empty, raising error if it is
        if self.is_empty():
            raise EmptyQueue("Queue is empty")
        else:
            # Return fronts value safely
            return self.data[self.front]

    def __len__(self) -> int:
        """Return the number of elements in the queue, Time complexity of O(1)"""
        return self.size

    def is_empty(self) -> bool:
        """Return whether the queue is empty, Time complexity of O(1)"""
        if self.size == 0:
            return True
        else:
            return False
```

#### Operations
The above class is just structure at this point, it is now time to show some operations run with the Queue, using: Insert, Delete, Search, Traversal.

##### Insert

```python
def circular_array_queue_insert(queue: CircularArrayQueue, count_of_e: int) -> None:
    """Insert x amount of ints into a CircularArrayQueue, print results"""
    for i in range(count_of_e):
        # Queues version of insert is enqueue
        queue.enqueue(i)
        print(f"First In: {queue.first()}\n"
              f"Size of Q: {queue.size}\n"
              f"Size of internal Array: {len(queue.data)}\n")



if __name__ == "__main__":
    circular_array_queue_insert(CircularArrayQueue(), 11)

```

After inserting the 11 element we can observe a few things. First, the value of first never changes, if we wanted to see deeper into this implementation, the only way would be to remove the first element. We can also observe that when the Queue reaches its max iniiliased size, it resizes itself to double its capsity.

##### Deletion

```python
def circular_array_queue_delete(queue: CircularArrayQueue, count_of_e: int) -> None:
    """Insert x amount of ints into a CircularArrayQueue, after each insert loop follow with a delete loop, print results"""
    for i in range(count_of_e):
        # Queues version of insert is enqueue
        queue.enqueue(i)
        print(f"First Value: {queue.first()}\n"
              f"Size of Q: {queue.size}\n"
              f"Size of internal Array: {len(queue.data)}\n")

    while True:
        # Loop through each element in the Queue, once empty throw EmptyQueue error, an example of using error as a definition of state
        try:
            # Queues version of delete, returns what was release
            element = queue.dequeue()
            print(f"First Value: {queue.first()}\n"
              f"Size of Q: {queue.size}\n"
              f"Size of internal Array: {len(queue.data)}\n"
              f"Released element: {element}\n")
        except EmptyQueue as e:
            print(e.args[0])
            break




if __name__ == "__main__":
    circular_array_queue_delete(CircularArrayQueue(), 11)
```

With mixing both enqueue with dequeue we can see the FIFO nature, new elements are added to the end, and when dequeue is called, whatever element that is first is release, moving the first pointer to the next in line. We cn also observe one of the issues with our implementaion of our CirularArrayQueue, even when the Queue becomes empty, the internal size of the Array stays the same. This creates a size completity of O(N) where N is the max amount of elements that have been in the Queue at one time at worse case * 2, and base case euqal to: N = n*2 || N = n. We can modifiy our implementaion to make the space completixty best case more eiffent, by resizing the Array by half if the Queue element count becomes 1/4 the size of the current Array, but in this case we will matintain what we have.

##### Search

```python
def circular_array_queue_search(queue: CircularArrayQueue, count_of_e: int, seek: int) -> None:
    """Insert x amount of ints into a CircularArrayQueue, search through the queue for seek item, print results, return -1 if not found"""
    # First populate our queue
    for i in range(count_of_e):
        queue.enqueue(i)

    # Destructive search, the queue is dismantled, but O(n) is maintained
    result = destructive_queue_search(queue, seek)
    print("Destructive Search returned: ", result)
    if result:
        print("Queue State: \n\n"
              f"First Value: {queue.first()}\n"
              f"Size of Q: {queue.size}\n"
              f"Size of internal Array: {len(queue.data)}\n")
    else:
        print("Queue State: \n\n"
              "First Value: None\n"
              f"Size of Q: {queue.size}\n"
              f"Size of internal Array: {len(queue.data)}\n")



def destructive_queue_search(queue: CircularArrayQueue, seek: int) -> bool:
    """Destructive Queue Search Algorithm, returns true if found, leaving found item at front of queue, false otherwise but queue is completely destroyed"""
    is_found = False

    try:
        while not is_found:
            print(f"Front: {queue.dequeue()}\n")
            if queue.front == seek:
                is_found = True
                return is_found
    except EmptyQueue:
        return is_found


if __name__ == "__main__":
    circular_array_queue_search(CircularArrayQueue(), 1000, 950)
```

With the use of a destructive search, we can obtain a O(n) worse case time complexity, this is the most efficient way to search through a queue, why maintaining the abstract data type structure. If we were to use the internal Array, we could run a number of different search algorithms, but it breaks down the abstraction. If we were to use a none destructive algorithm, it's possible to use a circular search, dequeue every element, if the seek element is found, mark it in a local variable, and enqueue every element back into place: this does give us the same worse case of O(n), but our best case is also pulled down to O(n), leaving every search call must iterate over every element even when its found. each method of searching is truly up to the use case, see more in reflection for more on this.

```python
def circular_array_queue_traversal(queue: CircularArrayQueue, count_of_e: int) -> None:
    """Insert x amount of ints into a CircularArrayQueue, traverse the whole queue, printing each slice"""
    # First populate our queue
    for i in range(count_of_e):
        queue.enqueue(i)
    # Run traversal
    queue_circular_traversal(queue)

    # run destructive traversal to assert same order is still maintained
    for i in range(queue.size):
        print(queue.dequeue())


def queue_circular_traversal(queue: CircularArrayQueue) -> None:
    """Traverse a queue in a circular manner, maintaining structure, time complexity of O(1)"""

    # Loop through each element, print it, then place back on the queue
    for i in range(queue.size):
        requeue_value = queue.dequeue()
        print(f"Value: {requeue_value}\n")
        queue.enqueue(requeue_value)





if __name__ == "__main__":
    circular_array_queue_traversal(CircularArrayQueue(), 11)
```

We can see how all of this now comes together, knowing that a ciruclar traversal will always need to hit every element, it creates a perfect way for us to walk through the Queue without causing any damange to the structure.

### Linked List
Linked Lists are another fundamental to Comp-Sci, and can be a great alternative to using Arrays or in Pythons case Lists. The high level of a Linked List is a collection of Nodes, each Node can have one or many references/ Edges to other Nodes. Linked Lists come a many different flavors, some of are: Signally Linked list(Node only point in one direction), Doubly Linked Lists(Nodes are Bi-Directional) and Circular Linked List(Tails Point to Head and Head to tails if Doubly Linked). The Nodes must be managed in some way, so normally a XTypeLinkedList is used to store methods and manage how the Nodes are handled.

Some use cases of Linked Lists include:
* Undo and redo features in code and text editors
* Back and forth operations for web browsers
* File and Directory systems
* Can be paired with or data structures for more completed algorithms

#### Implementation
A Linked List is a concrete data structure, meaning that it describes how data is physically organized and stored. Unlike an ADT, that only represents how behavior should look( E.G. class vs interface).

We can still use an ADT to define the minium expected behavior, using a List ADT, and provide a concrete implementation using a DoublyLinkedList.

A DoublyLinkedList follows the same idea as other implementations, a class holding two Node references for head and tail, but the Nodes within point to both the next Node and the last Node, creating a two lane highway.

```python
from typing import Any

class ListInterface:
    """A basic interface for lists and list like structures"""

    def insert(self, value: Any, index) -> None:
        """Insert value into the list, at index position"""
        pass

    def remove(self, index) -> None:
        """Remove value from the list, at index position"""
        pass

    def get(self, index: int) -> Any:
        """Return value at index position"""
        pass

    def set(self, index: int, value: Any) -> None:
        """Set value at index position"""
        pass

    def __len__(self) -> int:
        """Return length of list"""
        pass

    def is_empty(self) -> bool:
        """Return True if list is empty"""
        pass

    def clear(self) -> None:
        """Clear list"""
        pass

```

Now the List ADT can be seen, ill implement a DoublyLinkedList using it as a super class.

```python
class DoublyLinkedList(ListInterface):
    class ListEmptyException(Exception):
        """Exception raised when list is empty"""
        pass

    """A doubly linked list, implemented using lists ADT"""
    class _Node:
        """Doubly Linked List Node, points to last and next nodes"""
        def __init__(self,
                     value: Any = None,
                     last: DoublyLinkedList._Node = None,
                     next: DoublyLinkedList._Node = None
                     ) -> None:

            self.value = value
            self.last = last
            self.next = next

        def clean(self):
            self.last = None
            self.last = None
            self.next = None




    def __init__(self) -> None:
        # Initializes sentinels as head and tail(will not hold data, or effect size)
        self.head = self._Node()
        self.tail = self._Node()
        # Have the head and tail sentential point at each other, also points by default to out of bounds None
        self.head.next = self.tail
        self.tail.last = self.head

        self.size = 0

    def insert(self, value: Any, index) -> None:
        """Insert value into the list, at index position, relink Nodes, O(n) for search operation"""
        # Assign starting node as head
        node = self.head
        # internal index tracker
        i = 0

        if not self.is_empty():
            node = node.next

        # Loop through each node in the List, ending on tail node
        while node != self.tail:
            # if indices align
            if i == index:
                # Create new Node based on passed value, and point to last and next node
                new_node = self._Node(value, node, node.next)
                # Relink next Node last to new Node
                node.next.last = new_node
                # Relink last Node next to new Node
                node.next = new_node
                break
            else:
                # else get next Node
                node = node.next
                # Increment index tracker
                i =  i + 1
        # If index isn't found raise error
        if node == self.tail:
            raise IndexError("Index out of range")

    def remove(self, index: int) -> None:
        """Remove value at index position, relink Nodes, O(n) for search operation"""
        # Assign starting Node as head
        node = self.head
        #  Internal index tracker
        i = 0

        if self.is_empty():
            raise self.ListEmptyException("List is empty")
        else:
            node = node.next

        # Loop through each Node in the List, ending on tail Node
        while node != self.tail:
            if i == index:
                # Assign temp variable to node to remove
                del_node = node.next
                # Get next Node of Del_Node to relink it to del_nodes last
                del_node.next.last = node
                # Relink 1st Node to 3rd Node
                node.next = del_node.next
                # Remove all links to free the del_node
                del_node.clean()
                break
            else:
                # else get next Node
                node = node.next
                # Increment index tracker
                i =  i + 1
        # If index isn't found raise error
        if node == self.tail:
            raise IndexError("Index out of range")

    def add(self, value: Any, to_head: bool = True) -> None:
        """Add value to the front of list by default, if to_head is false adds to tail, relink Nodes, O(1) for insertion"""
        if to_head:
            # Create new Node, last pointing to head and next pointing to heads next
            new_node = self._Node(value, self.head, self.head.next)
            # moves heads next Nodes last pointer to new Node
            self.head.next.last = new_node
            # Moves heads next pointer to new Node
            self.head.next = new_node
        else:
            # Create new Node. last pointing to tails last, next pointing to tail
            new_node = self._Node(value, self.tail.last, self.tail)
            # Change tails last next pointer to new Node
            self.tail.last.next = new_node
            # Change tails last pointer to new Node
            self.tail.last = new_node
        # Increment size
        self.size += 1

    def pop(self, from_head: bool = True):
        """Pop Node from head of the list by default, if from head is false pops from tail, relink Nodes, O(1) for deletion"""

        if self.is_empty():
            raise self.ListEmptyException("List is empty")

        if from_head:
            # Logic for head
            # Assign temp reference for Node to remove
            del_node = self.head.next
            # Relink next Node.last of del_node to head Node
            del_node.next.last = self.head
            # Relink head.next to del_node.next
            self.head.next = del_node.next
            # Grab value before deletion
            del_value = del_node.value
            # Clean and remove del_node
            del_node.clean()
            # Decrement size
            self.size -= 1
        else:
            # Logic for tail
            # Assign a reference for Node to remove
            del_node = self.tail.last
            # Relink last.next Node from del_node to tail
            del_node.last.next = self.tail
            # Set tail pointer to del_node.last
            self.tail.last = del_node.last
            # Grab value before deletion
            del_value = del_node.value
            # Clean and release del_node
            del_node.clean()
            # Decrement size
            self.size -= 1
        # Return del_nodes value
        return del_value

    def get(self, index: int) -> Any:
        """Return value at index position, Loop has O(n) time complexity"""
        # Initialise search at head Node
        node = self.head
        # Internal index tracker
        i = 0

        if self.is_empty():
            raise self.ListEmptyException("List is empty")
        else:
            node = node.next

        # Loop through Nodes, if found return value of Node, if not raise index error
        while node != self.tail:
            # Check indices align
            if i == index:
                # return nodes value
                return node.value
            else:
                # Move to next Node
                node = node.next
                # Increment internal tracker
                i = i + 1
        if node == self.tail:
            # If index not found raise index error
            raise IndexError("Index out of range")
        else:
            # Just filler to Stop IDE freaking out, should not be reachable
            return None

    def set(self, index: int, value: Any) -> None:
        """Set value at index position, loop has O(n) time complexity"""
        # Initialise search at head Node
        node = self.head
        # Internal index tracker
        i = 0

        if self.is_empty():
            raise self.ListEmptyException("List is empty")
        else:
            node = node.next

        # Loop through Nodes, if found return value of Node, if not raise index error
        while node != self.tail:
            # Check indices align
            if i == index:
                # set node to new value
                node.value = value
            else:
                # Move to next Node
                node = node.next
                # Increment internal tracker
                i = i + 1
        if node == self.tail:
            # If index not found raise index error
            raise IndexError("Index out of range")
        else:
            # Just filler to Stop IDE freaking out, should not be reachable
            return None

    def clear(self) -> None:
        """Clear list, loop has O(n) time complexity"""
        # Set Node as head Node
        node = self.head

        # check if emtpy
        if self.is_empty():
            # Throw index error
            raise self.ListEmptyException("List is empty")

        # Loop from start to finish clearing all node data
        while node != self.tail:
            node_next = node.next
            node.clean()
            node = node_next
        # Relink head and tail to each other
        self.head.next = self.tail
        self.tail.last = self.head


    def is_empty(self) -> bool:
        """Return True if list is empty"""
        if self.size == 0:
            return True
        return False

    def __str__(self) -> str:
        list_str = "-Head-"
        node = self.head.next
        i = 0
        while node != self.tail:
            list_str += f"\n-Node_{i}::{node.value}-"
            node = node.next
        list_str += "\n-Tail-"
        return list_str




```
