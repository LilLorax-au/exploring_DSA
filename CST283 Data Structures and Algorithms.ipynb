### Queues
A Queue is one of the fundamental data structures used within computer science. The queue operates of the rule that what ever is placed in first must also come out first(FIFO). You can think of a queue simular to how you would line up to buy tickets for a show; whoever joined the line first, will be the first to be served, followed by who was after them, and continuing in this order, and only in this order.

Some more common use cases for Queue could include, but not limited to:
* Round Robin algorithms, in the planning and scheduling of tasks such as CPU processes and network load balancing
* Structured syntax handling, e.g. were brackets and parenthesis should open and close

#### Implementation
There is a few different ways Queues may be implemented, such as; Standard Queues, Circular Queues and the more distant Deque(_Pronounced deck_ also known as a double ended Queue). The example that is demonstrated below is a Circular Queue, implemented using a Dynamic Array.

First is the Queue as Abstract Data Type(AST), this is how we can define base behavior of queue, but can be modified for use case. Python doesn't support native interfaces, so a class with pass method is used to demonstration this. When going to use queue and queue-like structures, we will use this as a super/ parent class.

```python
from typing import Any

class EmptyQueue(Exception):
    """Exception raised when queue is empty"""
    pass


class QueueInterface:
    def enqueue(self, element: Any) -> None:
        """Insert/Add an element to the back of the queue"""
        pass
    def dequeue(self) -> Any:
        """Remove and return the element from the front of the queue"""
        pass
    def first(self) -> Any:
        """Safe return the first element in the queue"""
        pass
    def __len__(self) -> int:
        """Return the number of elements in the queue"""
        pass
    def is_empty(self) -> bool:
        """Return whether the queue is empty"""
        pass
```

Now we will implement the CircularArrayQueue using the above class

```python
from typing import Any

class CircularArrayQueue(QueueInterface):
    DEF_CAP: int = 10

    def __init__(self):
        """Initialize an empty queue"""
        # Define the inner structure, using a list as a Dynamic Array
        self.data: list = [None] * self.DEF_CAP
        self.size: int = 0
        self.front: int = 0


    def enqueue(self, element: Any) -> None:
        """Insert/Add an element to the back of the queue, has a time complexity of O(1) and maintains an atomized time complexity of O(1) on resize, only if resize doubles capacity"""
        # First check if the Array is full
        if self.size == len(self.data):
            # If at capacity, call resize
            self.resize(len(self.data) * 2)
        # Avail points to the back of the queue, this is found from: queues front index + size of the queue modulus the total size of array
        # E.G. if front sits at index 9, queue size of 4, and array size of 10, then avail points to index 3, (9 + 4) % 10 = 3
        # We do this to imutate the Array as if it were an ouroboros, its tail connecting to it head
        avail = (self.front + self.size) % len(self.data)
        # Now just add the element to where the new tail position sits
        self.data[avail] = element
        # Increment size
        self.size += 1


    def dequeue(self) -> Any:
        """Remove and return the element from the front of the queue, use case maintains a O(1) time complexity"""
        if self.is_empty():
            raise EmptyQueue("Queue is empty")

        # Set value to maintain a reference for the current front of the queue
        value = self.data[self.front]
        # Remove fronts value
        self.data[self.front] = None
        # Using simular process to find the back of the queue, we find the new front
        # Maintaining the circular structure, we use the last index of front, increment 1, and modulus the length of the Array,
        # E.G. if front index is 9, and Array length is 10, we get the new index of 0, (9 + 1) % 10 = 0
        self.front = (self.front + 1) % len(self.data)
        # decrement by 1
        self.size -= 1
        # return maintained old front
        return value

    def resize(self, cap: int) -> None:
        """Resize the queue to the specified capacity, maintains worse case time complexity of O(n)"""
        # Take a copy of the old data
        old = self.data
        # Create a new fixed sized list to mimic a Array, based on passed parameter
        self.data = [None] * cap
        # Grab a reference to the queues front
        walk = self.front

        # Iterate over all the old elements to place into new Array
        for k in range(self.size):
            # Place old front to the beginning of new Array
            self.data[k] = old[walk]
            # Find the next element in the old array using modulus arithmetic
            walk = (1 + walk) % len(old)
        # Reset front value to match new Array
        self.front = 0
        # Old Array leaves scope and is marked for garbage collection

    def first(self) -> Any:
        """Safe return the first element in the queue, Time complexity of O(1)"""
        # First check is Queue is empty, raising error if it is
        if self.is_empty():
            raise EmptyQueue("Queue is empty")
        else:
            # Return fronts value safely
            return self.data[self.front]

    def __len__(self) -> int:
        """Return the number of elements in the queue, Time complexity of O(1)"""
        return self.size

    def is_empty(self) -> bool:
        """Return whether the queue is empty, Time complexity of O(1)"""
        if self.size == 0:
            return True
        else:
            return False
```
